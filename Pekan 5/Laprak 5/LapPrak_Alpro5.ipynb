## 2.1 Dasar Teori
1. <b>Perulangan `for` pada Python</b><br> 
Perulangan `for` pada Python digunakan untuk mengulang suatu blok kode berdasarkan urutan elemen atau rentang nilai yang ditentukan. Berbeda dengan bahasa pemrograman C/C++ atau Java yang berbasis variabel indeks tercatat secara langsung, perulangan `for` pada Python bekerja secara iteratif berdasarkan objek.<br><br>
Sintaks umum perulangan `for`:
```python
for variabel in iterable:
    # blok kode yang dieksekusi
```

2. <b>Fungsi `range()`</b> <br>
Fungsi range() sering digunakan bersamaan dengan for untuk menghasilkan deret bilangan bulat. Fungsi ini dapat menerima hingga tiga argumen:<br>
    - `range(stop)`: Menghasilkan deret angka dari `0` hingga `stop - 1`.
    - `range(start, stop)`: Menghasilkan deret angka dari `start` hingga `stop - 1`.
    - `range(start, stop, step)`: Menghasilkan deret angka dari `start` hingga `stop - 1` dengan penambahan/pengurangan sebesar `step`. 

3. <b>Perulangan Bersarang (Nested Loop)</b><br>
Nested loop terjadi ketika terdapat blok perulangan di dalam blok perulangan lainnya. Dalam nested loop, perulangan tingkat dalam (inner loop) akan dieksekusi hingga selesai untuk setiap satu iterasi dari perulangan tingkat luar (outer loop). Konsep ini penting dalam pembuatan pola dua dimensi (baris dan kolom) serta pemrosesan matriks.<br><br>
Sintaks umum nested loop:
```python
for i in range(baris):
    for j in range(kolom):
        # blok kode dieksekusi di setiap sel matriks
```

### 2.2.3 Program Perulangan for3 (`Perulangan_for3_2611533025.py`)
</div>

````python
# Buat file dengan nama Perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3025 = int(input("Masukkan jumlah perulangan: "))

jumlah_3025 = 0
for i_3025 in range(1, ulang_3025 + 1):
    print(i_3025, end=" ")
    jumlah_3025 = jumlah_3025 + i_3025

    if i_3025 < ulang_3025:
        print("+", end=" ")
    else:
        print(" = ", jumlah_3025, end=" ")
print()
print("jumlah =", jumlah_3025)
````

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

Penjelasan:
* **Baris 1–4:** Komentar instruksi pemuatan variabel.
* **Baris 6:** `ulang_3025 = int(input("Masukkan jumlah perulangan: "))` — Menerima input integer jumlah angka yang akan dijumlahkan.
* **Baris 8:** `jumlah_3025 = 0` — Inisialisasi variabel akumulator `jumlah_3025` dengan nilai awal `0`.
* **Baris 9:** `for i_3025 in range(1, ulang_3025 + 1):` — Perulangan dari `1` hingga `ulang_3025`.
* **Baris 10:** `print(i_3025, end=" ")` — Mencetak angka iterasi saat ini tanpa membuat baris baru.
* **Baris 11:** `jumlah_3025 = jumlah_3025 + i_3025` — Menambahkan nilai `i_3025` ke variabel akumulator.
* **Baris 13:** `if i_3025 < ulang_3025:` — Mengecek apakah iterasi masih berada di tengah (bukan angka terakhir).
* **Baris 14:** `print("+", end=" ")` — Jika bukan angka terakhir, cetak simbol `+`.
* **Baris 15–16:** `else: print(" = ", jumlah_3025, end=" ")` — Jika angka terakhir, cetak simbol `=` beserta hasil akumulasinya.
* **Baris 17:** `print()` — Pindah ke baris baru.
* **Baris 18:** `print("jumlah =", jumlah_3025)` — Menampilkan total penjumlahan secara keseluruhan.

### 2.2.5 Nested for1 (`Nested_for1_2611533025.py`)

````python
# Buat file dengan nama Nested_for1_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3025 = int(input("Masukkan nilai batas: "))
for line_3025 in range(1, batas_3025 + 1):
    for j_3025 in range(1, (-1 * line_3025 + batas_3025) + 1):
        print(".", end=" ")
    print(line_3025)
````

### 2.2.8 Program Nested for4 (`Nested_for4_2611533025.py`)


</div>

````python
# Buat file dengan nama Nested_for4_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_3025 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3025 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3025 = tinggi_3025
    c_3025 = a_3025 
    lebar_3025 = (2 * tinggi_3025) - 2

    for i_3025 in range(1, tinggi_3025 + 1):
        b_3025 = c_3025 + 1

        for j_3025 in range(1, lebar_3025 + 1):

            #Baris atas dan bawah
            if i_3025 == 1 or i_3025 == tinggi_3025:
                if j_3025 == 1 or j_3025 == lebar_3025:
                    print("#", end="")
                else:
                    print("=", end="")

            #Baris isi
            else:
                if j_3025 == 1 or j_3025 == lebar_3025:
                    print("|", end="")
                else:
                    if j_3025 == c_3025:
                        print("<", end="")
                    elif j_3025 == b_3025:
                        print(">", end="")
                    elif j_3025 == (lebar_3025 - c_3025):
                        print("<", end="")
                    elif j_3025 == (lebar_3025 - c_3025 + 1):
                        print(">", end="")
                    elif j_3025 > b_3025 and j_3025 < (lebar_3025 - c_3025):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # logika asli Java
        a_3025 -= 2

        if a_3025 <= 0:
            c_3025 = (-a_3025) + 2
        else:
            c_3025 = a_3025
````

<div style=" font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

Penjelasan:
* **Baris 6:** Menerima input tinggi pola yang diinginkan pengguna.
* **Baris 8–9:** Memvalidasi apakah tinggi merupakan bilangan genap; jika ganjil, tampilkan pesan peringatan.
* **Baris 11–13:** Mempersiapkan variabel bantu `a_3025`, `c_3025`, dan menghitung `lebar_3025` pola.
* **Baris 15–16:** *Outer loop* mengiterasi baris `i_3025` dan menghitung posisi pembatas tengah `b_3025`.
* **Baris 18:** *Inner loop* mengiterasi kolom `j_3025` dari `1` sampai `lebar_3025`.
* **Baris 21–25:** Logika pembentukan bingkai atas dan bawah (ujung memakai `#`, sisanya memakai `=`).
* **Baris 28–43:** Logika isi pola:
  * Mengisi batas kiri/kanan dengan `|`.
  * Menempatkan simbol `<` dan `>` berdasarkan perhitungan variabel `c_3025` dan `b_3025` untuk membentuk kontur jam pasir.
  * Menisi titik `.` di area dalam jam pasir dan spasi di luar jam pasir.
* **Baris 45:** berpindah ke baris baru.
* **Baris 48–53:** Mengubah nilai pergeseran `a_3025` dan `c_3025` untuk membalikkan kontur jam pasir setelah melewati titik tengah.

````python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3025 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ---------- Bingkai atas ----------
print("#", end="")
for pagar_3025 in range(4 * n_3025 + 5):
    print("=", end="")
print("#", end="")
print()

# ---------- Fase 1: jam pasir atas (N turun s.d. 1) ----------
for baris_3025 in range(n_3025, 0, -1):
    print("| ", end="")
    for spasi_kiri_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    for angka_3025 in range(baris_3025, 0, -1):
        print(angka_3025, end=" ")
    print("<*>", end="")
    for angka_3025 in range(1, baris_3025 + 1):
        print(" ", end="")
        print(angka_3025, end="")
    for spasi_kanan_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    print(" |", end="")
    print()

# ---------- Fase 2: poros titik pusat ----------
print("|", end="")
for spasi_kiri_3025 in range(2 * n_3025 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_kanan_3025 in range(2 * n_3025 + 1):
    print(" ", end="")
print("|", end="")
print()

# ---------- Fase 3: jam pasir bawah (1 naik s.d. N) ----------
for baris_3025 in range(1, n_3025 + 1):
    print("| ", end="")
    for spasi_kiri_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    for angka_3025 in range(baris_3025, 0, -1):
        print(angka_3025, end=" ")
    print("<*>", end="")
    for angka_3025 in range(1, baris_3025 + 1):
        print(" ", end="")
        print(angka_3025, end="")
    for spasi_kanan_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    print(" |", end="")
    print()

# ---------- Bingkai bawah ----------
print("#", end="")
for pagar_3025 in range(4 * n_3025 + 5):
    print("=", end="")
print("#", end="")
print()
````
