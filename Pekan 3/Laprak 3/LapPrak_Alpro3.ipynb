## 2.1 Dasar Teori
1. Operator Aritmatika: Simbol matematika yang digunakan untuk melakukan operasi hitung dasar, seperti penjumlahan (`+`), pengurangan (`-`), perkalian (`*`), pembagian (`/`), pembagian bulat (`//`), sisa bagi/modulus (`%`), dan perpangkatan (`**`).

2. Operator Perbandingan (Relasional): Operator yang membandingkan dua nilai dan menghasilkan nilai boolean (`True` atau `False`). Meliputi lebih besar (`>`), lebih kecil (`<`), lebih besar sama dengan (`>=`), lebih kecil sama dengan (`<=`), sama dengan (`==`), dan tidak sama dengan (`!=`).

3. Operator Logika: Digunakan untuk membandingkan dua ekspresi boolean. Terdiri dari `and` (bernilai `True` jika keduanya `True`), `or` (bernilai `True` jika salah satu `True`), dan `not` (membalikkan nilai boolean).

4. Operator Assignment (Penugasan): Digunakan untuk memberikan atau menguji nilai ke dalam variabel. Operator gabungan (seperti `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`) melakukan operasi matematika sekaligus memperbarui nilai variabel tersebut.

5. Operator Keanggotaan & Identitas:

    - Keanggotaan (`in`, `not in`): Memeriksa apakah suatu nilai ada di dalam sebuah urutan/koleksi data (seperti List atau String).

    - Identitas (`is`, `is not`): Membandingkan apakah dua variabel merujuk pada objek/lokasi memori yang sama (berbeda dengan == yang hanya membandingkan nilainya).

6. Operator Bitwise: Operator yang memanipulasi bit angka secara langsung pada tingkat biner (0 dan 1). Terdiri dari AND (`&`), OR (`|`), XOR (`^`), NOT (`~`), Left Shift (`<<`), dan Right Shift (`>>`).

### 2.2.1 Program Aritmatika (`Aritmatika_2611533025.py`)

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_3025 = int(input("Input angka-1: "))
angka2_3025 = int(input("Input angka-2: "))

# Penjumlahan
hasil_3025 = angka1_3025 + angka2_3025
print("\nOperator Penjumlahan:")
print("Hasil =", hasil_3025)

# Pengurangan
hasil_3025 = angka1_3025 - angka2_3025
print("\nOperator Pengurangan:")
print("Hasil =", hasil_3025)

# Perkalian
hasil_3025 = angka1_3025 * angka2_3025
print("\nOperator Perkalian:")
print("Hasil =", hasil_3025)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_3025 != 0:
    hasil_3025 = angka1_3025 / angka2_3025
    print("\nOperator Pembagian:")
    print("Hasil =", hasil_3025)

    hasil_3025 = angka1_3025 // angka2_3025
    print("\nOperator Pembagian Bulat:")
    print("Hasil =", hasil_3025)

    hasil_3025 = angka1_3025 % angka2_3025
    print("\nOperator Sisa Bagi:")
    print("Hasil =", hasil_3025)
else:
    print("Angka kedua tidak boleh bernilai 0.")

hasil_3025 = angka1_3025 ** angka2_3025
print("\nOperator Pangkat:")
print("Hasil =", hasil_3025)
```

### 2.2.3 Program Logika (`Logika_2611533025.py`)

```python
# Masukkan nilai boolean
#input tidak peka terhadap huruf besar dan kecil
a1_3025 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_3025 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1_3025 =", a1_3025)
print("A2_3025 =", a2_3025)

# Konjungsi: bernilai True jika keduanya True
hasil_3025 = a1_3025 and a2_3025
print("\nOperator Konjungsi (AND):")
print("A1_3025 and A2_3025:", hasil_3025)

# Disjungsi: bernilai True jika salah satunya True
hasil_3025 = a1_3025 or a2_3025
print("\nDisjungsi (OR)")
print("A1_3025 or A2_3025:", hasil_3025)

# Negasi A1: membalik nilai A1
hasil_3025 = not a1_3025
print("\nNegasi A1_3025 (NOT)")
print("not A1_3025:", hasil_3025)

# Negasi A2: membalik nilai A2
hasil_3025 = not a2_3025
print("\nNegasi A2_3025 (NOT)")
print("not A2_3025:", hasil_3025)

# XOR: bernilai True jika kedua nilai berbeda
hasil_3025 = a1_3025 != a2_3025
print("\nDisjungsi Eksklusif (XOR)")
print("A1_3025 XOR A2_3025:", hasil_3025)
```

### 2.2.4 Program Assigment (`Assigment_2611533025.py`)
</div>

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_3025 = int(input("Input angka-1: "))
angka2_3025 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_3025)
print("Nilai awal angka2 =", angka2_3025)

# Assigment biasa
hasil_3025 = angka1_3025
print("\nHasil assigment biasa (=)")
print("Hasil =", hasil_3025)

# Assigment penambahan
hasil_3025 = angka1_3025 
hasil_3025 += angka2_3025
print("\nAssigment penambahan (+=)")
print("Hasil =", hasil_3025)

# Assigment pengurangan
hasil_3025 = angka1_3025
hasil_3025 -= angka2_3025
print("\nAssigment pengurangan (-=)")
print("Hasil =", hasil_3025)

# Assigment perkalian
hasil_3025 = angka1_3025
hasil_3025 *= angka2_3025
print("\nAssigment perkalian (*=)")
print("Hasil =", hasil_3025)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_3025 != 0:
    hasil_3025 = angka1_3025
    hasil_3025 /= angka2_3025
    print("\nAssigment pembagian (/=)")
    print("Hasil =", hasil_3025)
    # Operator tambahan
    hasil_3025 = angka1_3025
    hasil_3025 //= angka2_3025
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =", hasil_3025)

    hasil_3025 = angka1_3025
    hasil_3025 %= angka2_3025
    print("\nAssigment sisa bagi (%=)")
    print("Hasil =", hasil_3025)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assigment perpangkatan
hasil_3025 = angka1_3025
hasil_3025 **= angka2_3025
print("\nAssigment perpangkatan (**=)")
print("Hasil =", hasil_3025)
```

### 2.2.5 Program Lainnya (`Lainnya_2611533025.py`)
</div>

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("======================================")
print("1. OPERATOR KEANGGOTAAN")
print("======================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_3025 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_3025 = [int(angka.strip()) for angka in input_data_3025.split(',')]

nilai_dicari_3025 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_3025 = nilai_dicari_3025 in data_3025
print("\nOperator keanggotaan IN")
print(nilai_dicari_3025,"in",data_3025,"=",hasil_3025)

# Operator not in
hasil_3025 = nilai_dicari_3025 not in data_3025
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_3025,"not in",data_3025,"=",hasil_3025)

print("======================================")
print("2. OPERATOR IDENTIAS")
print("======================================")

# objek1 menggunakan list dari input perngguna
object1_3025 = data_3025

# objek2 merujuk pada objek yang sama dengan objek1
object2_3025 = object1_3025

# objek3 memiliki isi sama, tetapi merupakan objek baru
object3_3025 = data_3025.copy()

print("object1 =",object1_3025)
print("object2 =",object2_3025)
print("object3 =",object3_3025)

# Operator is
hasil_3025 = object1_3025 is object2_3025
print("\nOperator identitas IS")
print("objek1 is objek2 =",hasil_3025)

# Operator is not
hasil_3025 = object1_3025 is not object2_3025
print("\nOperator identitas IS NOT")
print("objek1 is not objek2 =",hasil_3025)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =",object1_3025 is object3_3025)
print("objek1 == objek3 =",object1_3025 == object3_3025)
```

`object2_3025` merujuk ke memori yang sama dengan `object1_3025`, sedangkan `object3_3025` menyalin isinya saja ke lokasi memori baru.

5. Operator Identitas IS


`hasil_3025 = object1_3025 is object2_3025`

Operator `is` mengecek apakah dua variabel berada di lokasi memori yang sama. Menghasilkan `True` untuk `object1_3025` dan `object2_3025`.

6. Operator Identitas IS NOT


`hasil_3025 = object1_3025 is not object2_3025`

Operator `is not` mengecek apakah dua variabel berada di lokasi memori yang berbeda.

7. Perbandingan Nilai vs Lokasi Memori


`object1_3025 is object3_3025` menghasilkan `False` karena beda lokasi memori.


`object1_3025 == object3_3025` menghasilkan `True` karena isinya sama persis.

### 2.2.6 Program Bitwise (`Bitwise_2611533025.py`)

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_3025 = int(input("Masukkan angka bitwise-1: "))
angka2_3025 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_3025,"| biner =",bin(angka1_3025))
print("angka1 =",angka2_3025,"| biner =",bin(angka2_3025))

# Bitwise AND
hasil_3025 = angka1_3025 & angka2_3025
print("\nBitwise AND (&)")
print(angka1_3025,"&",angka2_3025,hasil_3025)
print("Biner hasil =",bin(hasil_3025))
print("Biner hasil (8 bit) =",format(hasil_3025,"08b"))

# Bitwise OR
hasil_3025 = angka1_3025 | angka2_3025
print("\nBitwise OR (|)")
print(angka1_3025,"|",angka2_3025,hasil_3025)
print("Biner hasil =",bin(hasil_3025))
print("Biner hasil (8 bit) =",format(hasil_3025,"08b"))

# Bitwise XOR
hasil_3025 = angka1_3025 ^ angka2_3025
print("\nBitwise XOR (^)")
print(angka1_3025,"^",angka2_3025,hasil_3025)
print("Biner hasil =",bin(hasil_3025))
print("Biner hasil (8 bit) =",format(hasil_3025,"08b"))

# Bitwise NOT
hasil_3025 = ~angka1_3025
print("\nBitwise NOT (~)")
print(angka1_3025,"~",angka2_3025,hasil_3025)
print("Biner hasil =",bin(hasil_3025))
print("Biner hasil (8 bit) =",format(hasil_3025,"08b"))

# Bitwise geser kiri
jumlah_geser_3025 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3025 = angka1_3025 << jumlah_geser_3025
print("\nBitwise geser kiri (<<)")
print(angka1_3025,"<<",jumlah_geser_3025,"=",hasil_3025)
print("Biner hasil =",bin(hasil_3025))
print("Biner hasil (8 bit) =",format(hasil_3025,"08b"))

# Bitwise geser kanan
hasil_3025 = angka1_3025 >> jumlah_geser_3025
print("\nBitwise geser kanan (>>)")
print(angka1_3025,">>",jumlah_geser_3025,"=",hasil_3025)
print("Biner hasil =",bin(hasil_3025))
print("Biner hasil (8 bit) =",format(hasil_3025,"08b"))
```

Penjelasan:
1. Membuat input angka desimal & menampilkan biner

`bin(angka1_3025)`

Program meminta dua angka desimal, kemudian fungsi `bin()` mengonversi angka desimal tersebut menjadi format biner.

2. Bitwise AND (&)

`hasil_3025 = angka1_3025 & angka2_3025`

Operator `&` membandingkan bit demi bit dari kedua angka. Bit hasil akan bernilai 1 jika kedua bit bernilai 1.

3. Bitwise OR (|)

`hasil_3025 = angka1_3025 | angka2_3025`

Operator `|` membandingkan bit demi bit, bernilai 1 jika minimal salah satu bit bernilai 1.

4. Bitwise XOR (^)

`hasil_3025 = angka1_3025 ^ angka2_3025` 

Operator `^` membandingkan bit demi bit, bernilai 1 jika bit-bitnya berbeda (satu bernilai 0 dan lainnya bernilai 1).

5. Bitwise NOT (~)

`hasil_3025 = ~angka1_3025`

Operator `~` membalikkan semua bit angka (sistem komplementer dua), menghasilkan rumus `-(x + 1)`.

6. Bitwise Geser Kiri (<<)

`hasil_3025 = angka1_3025 << jumlah_geser_3025`

Operator `<<` menggeser posisi bit ke kiri sebanyak `jumlah_geser_3025` langkah (sama dengan mengalikan angka dengan $2^n$).

7. Bitwise Geser Kanan (>>)

`hasil_3025 = angka1_3025 >> jumlah_geser_3025`

Operator `>>` menggeser posisi bit ke kanan sebanyak `jumlah_geser_3025` langkah (sama dengan membagi bulat angka dengan $2^n$).
