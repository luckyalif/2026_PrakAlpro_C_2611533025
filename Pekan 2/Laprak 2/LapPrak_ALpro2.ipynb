## 2.1 Dasar Teori

1. Tipe Data Boolean
Tipe data Boolean (`bool`) merupakan tipe data paling dasar dalam logika pemrograman yang hanya memiliki dua nilai mutlak, yaitu `True` (benar) dan `False` (salah) (Python Software Foundation, 2026).

    - Pengujian Kondisi: Nilai boolean umumnya dihasilkan dari operasi perbandingan menggunakan operator seperti `>=`, `<=`, `==`, atau `!=`.

    - Ekspresi Logika: Operasi logika pada boolean dapat digabungkan menggunakan operator logika seperti `and`, `or`, dan `not` untuk menent`ukan keputusan pada alur program.

    - Pengendalian Alur (Control Flow): Digunakan sebagai syarat utama dalam struktur percabangan (`if-else`) guna menentukan blok kode mana yang akan dieksekusi oleh sistem.

2. Tipe Data Input dan Konversi Tipe Data (Type Casting)
Input merupakan mekanisme interaksi antara pengguna dan program untuk memasukkan data secara dinamis ke dalam memori (Microsoft, 2026).

    - Fungsi `input()`: Digunakan untuk menerima masukan teks dari pengguna melalui terminal console.

    - Sifat Luaran Teks (String Default): Semua nilai yang diambil melalui fungsi `input()` secara otomatis disimpan sebagai tipe data string (`str`), meskipun penguna memasukkan karakter angka.

    - Konversi Tipe Data (Type Casting): Karena luaran `input()` berupa string, diperlukan proses konversi tipe data secara eksplisit menggunakan fungsi seperti `int()` atau `float()` agar data angka tersebut dapat dihitung dalam operasi aritmetika.

3. Tipe Data Konstanta (Constant)
Secara bawaan (native), Python tidak memiliki fitur penentu variabel yang nilainya benar-benar terkunci mutlak sebagai konstanta seperti bahasa pemrograman lain.

    - Konvensi Penulisan: Secara tradisional, variabel yang bernilai tetap ditandai dengan penulisan nama variabel menggunakan huruf kapital secara keseluruhan (contoh: `PI`).

    - Penggunaan `typing.Final`: Python menyediakan modul `Final` dari pustaka bawaan `typing` untuk menegaskan secara eksplisit bahwa variabel tersebut bersifat konstan dan nilainya tidak boleh diubah sepanjang program berjalan (Python Software Foundation, 2026).

    - Keamanan Kode: Menjaga nilai penting—seperti konstanta matematika atau konfigurasi sistem—agar tidak tidak sengaja tertimpa (overwrite) oleh baris kode lain.

4. Tipe Data Numerik
Tipe data numerik digunakan untuk menyimpan data bertipe angka. Python mendukung tiga jenis tipe data numerik utama:

    - `Integer` (`int`): Tipe data yang menyimpan bilangan bulat positif atau negatif tanpa koma/desimal (contoh: `100`, `-5`).

    - `Floating-Point` (`float`): Tipe data yang menyimpan bilangan desimal atau pecahan yang ditandai dengan tanda titik (.) (contoh: `10.2345`).

    - `Complex` (`complex`): Tipe data yang menyimpan bilangan kompleks atau imajiner, di mana bagian imajiner ditandai dengan akhiran huruf `j` (contoh: `100 + 3j`).

5. Tipe Data String
Tipe data `String` (`str`) merupakan sekumpulan karakter yang dibentuk oleh susunan huruf, angka, simbol, atau spasi (Python Software Foundation, 2026).

    - Petik Tunggal (`' '`) / Petik Ganda (`" "`): Digunakan untuk mendefinisikan kalimat atau teks dalam satu baris tunggal.

    - Petik Tiga (`""" """`): Digunakan untuk mendefinisikan teks string yang spans lebih dari satu baris (multiline string), di mana format baris baru dan spasi indentasi di dalamnya tetap dipertahankan.

    - Identifikasi Tipe Data (`type()`): Fungsi bawaan `type()` dapat digunakan untuk memeriksa jenis tipe data dari suatu variabel yang sedang diolah.

### 2.2.1 Program Numeric_2611531021

```python
#create a variable with integer value
a_3025=100
print("The type of variabel having value", a_3025, "is",type(a_3025))

#create a variabel with float value.
b_3025=10.2345
print("The type of variabel having value", b_3025, "is",type(b_3025))

#create a variabel eith complex value.
c_3025=100+3j
print("The type of variabel having value", c_3025, "is",type(c_3025))
```

### 2.2.5 Program Konstanta_2611531021
</div>

```python
# Program ini menggunakan konstanta untuk menghitun luas lingkaran

from typing import Final
PI: Final = 3.14
print ("pi: %f" % (PI))
jari_3025 = float(input('masukkan nilai jari jari: '))
luas_3025 = PI * jari_3025 * jari_3025
print ("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3025, luas_3025))
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Mengimpor entitas `Final` dari modul `typing`.

2. Deklarasi `PI: Final = 3.14` sebagai variabel konstanta berjenis Float.

3. Cetak nilai `PI` dengan format bilangan desimal/float (`"pi: %f" % (PI)`).

4. Mengambil masukan nilai jari-jari dari pengguna dengan `input()`, dikonversi ke Float, lalu disimpan pada variabel `jari_3025`.
