## 2.1 Dasar Teori
1. <b>If Tunggal:</b> Digunakan saat program hanya perlu mengeksekusi aksi jika suatu kondisi terpenuhi, dan mengabaikannya jika kondisi bernilai `False`.

2. <b>If-Else:</b> Struktur percabangan dua jalur. Jika kondisi bernilai `True`, blok `if` dieksekusi; jika `False`, blok `else` yang dijalankan.

3. <b>If-Elif-Else:</b> Struktur percabangan multi-jalur mutlak (mutually exclusive). Evaluasi dilakukan secara berurutan dari atas ke bawah. Ketika salah satu kondisi bernilai `True`, blok kode tersebut dieksekusi dan seluruh sisa blok `elif`/`else` di bawahnya diabaikan.

4. <b>Multi-IF Terpisah:</b> Struktur yang menggunakan beberapa blok `if` mandiri secara berurutan. Setiap blok `if` dievaluasi secara independen sehingga memungkinkan pengakumulasian nilai (seperti akumulasi diskon) apabila lebih dari satu kondisi terpenuhi secara bersamaan.

5. <b>Match-Case:</b> Fitur structural pattern matching yang diperkenalkan pada Python 3.10. Digunakan untuk menggantikan rantai `if-elif` panjang pada pemilihan nilai diskrit, di mana `case _` berfungsi sebagai wildcard (pilihan default).

## 2.2 Kode Program
### 2.2.1 Program IF 1 (`if1_2611533025.py`)

</div>

````python
# Buat file dengan nama if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_3025 = float(input("Input IPK Anda = "))

if ipk_3025 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3025))

print("Program Selesai")
````

2. Percabangan IF Tunggal<br>
`if ipk_3025 > 2.75:` <br>
&emsp;`print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3025))` <br>
Program mengevaluasi apakah `ipk_3025` lebih besar dari 2.75. Jika bernilai `True`, pesan kelulusan dicetak. Jika `False`, blok `if` dilewati.

### 2.2.2 Program IF 2 (`if2_2611533025.py`)

````python
# Buat file dengan nama if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_3025 = float(input("Input IPK Anda = "))

if ipk_3025 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3025))
else:  
    print("Anda Tidak Lulus")
print("Program Selesai")
````

`if ipk_3025 > 2.75:` <br>
`else:` <br>
Jika kondisi `ipk_3025 > 2.75` bernilai `True`, blok `if` dieksekusi. Jika bernilai `False` (IPK $\le$ 2.75), kontrol dialihkan ke blok `else` untuk mencetak "Anda Tidak Lulus".

### 2.2.3 Program Multi-IF 1 (`multi_if1_2611533025.py`)

</div>

````python
# Buat file dengan nama multi_if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_3025 = int(input("Input Umur Anda: "))
sim_3025 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_3025 >= 17 and sim_3025 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")
if umur_3025 >= 17 and sim_3025 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
if umur_3025 < 17 and sim_3025 == "y":
    print("Anda Belum Cukup Umur punya SIM")
if umur_3025 < 17 and sim_3025 != "y":
    print("Anda Belum Cukup Umur bawa motor")
````

<div style="height:1123px; font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

Penjelasan:
1. Independensi Multi-IF Terpisah<br>
`if umur_3025 >= 17 and sim_3025 == "y":` <br>
`if umur_3025 >= 17 and sim_3025 != "y":` <br>
`if umur_3025 < 17 and sim_3025 == "y":` <br>
`if umur_3025 < 17 and sim_3025 != "y":` <br><br>
Setiap pernyataan `if` berdiri sendiri. Interpreter Python mengevaluasi keempat kondisi tersebut secara berurutan tanpa saling mengabaikan.

### 2.2.5 Program Multi-IF 2 (`multi_if2_2611533025.py`)


</div>

````python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3025 = int(input("Input Total Belanja Anda = Masukkan total belanja (RP): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3025 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_3025 = input_member_3025 in ["y", "ya"]

# Input status promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3025 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_Valid_3025 = input_promo_3025 in ['y', 'ya']

total_diskon_persen_3025 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3025 > 100000:
    total_diskon_persen_3025 += 10  # Diskon belanja besar

if is_member_3025:
    total_diskon_persen_3025 += 5  # Diskon member

if kode_promo_Valid_3025:
    total_diskon_persen_3025 += 15  # Diskon voucher

# menghitung nominal diskon dan total bayar
nominal_diskon_3025 = total_belanja_3025 * (total_diskon_persen_3025 / 100)
total_bayar_3025 = total_belanja_3025 - nominal_diskon_3025

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon: {total_diskon_persen_3025}% (Rp {nominal_diskon_3025:,.0f})")
print(f"Total Bayar: Rp {total_bayar_3025:,.0f}")

print (f"Total diskon yang anda dapatkan adalah {total_diskon_persen_3025}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
````

<div style="height:1123px; font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

Penjelasan:
1. String Handling dan Pengecekan Keanggotaan List<br>
`is_member_3025 = input_member_3025 in ["y", "ya"]` <br>
Menggunakan `.strip().lower()` untuk membersihkan spasi dan mengonversi masukan ke huruf kecil, lalu mengecek apakah input berada dalam list `["y", "ya"]`.

2. Penumpukan Diskon Akumulatif<br>
`if total_belanja_3025 > 100000:` <br>
&emsp;`total_diskon_persen_3025 += 10` <br>
`if is_member_3025:` <br>
&emsp;`total_diskon_persen_3025 += 5` <br>
`if kode_promo_Valid_3025:` <br>
&emsp;`total_diskon_persen_3025 += 15` <br>
Karena menggunakan blok `if` terpisah, variabel `total_diskon_persen_3025` akan ditambahkan nilainya secara bertahap sesuai dengan setiap syarat yang terpenuhi.
</div>

### 2.2.7 Program Match Case 1 (`match_case1_2611533025.py`)


</div>

````python
# Buat file dengan nama match_case_nim.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konvesi angka menjadi nama bulan

bulan_3025 = int(input("Inputkan nomor bulan (1-12): "))

match bulan_3025:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
````

<div style="height:1123px; font-family: 'Times New Roman', Times, serif; font-size: 16px; text-align: justify;">

Penjelasan:
1. Pemadanan Pola dengan Match-Case<br>
`match bulan_3025:` <br>
Mencocokkan nilai variabel `bulan_3025` dengan deretan opsi `case`.

2. Wildcard Case Default<br>
`case _:` <br>
Simbol `_` bertindak sebagai penampung default jika input angka pengguna berada di luar rentang 1–12.

````python
"""
Tugas 4 - Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
Nama   : Lucky Alif Pratama
NIM    : 2611533025
Materi : if tunggal, if-else, if-elif-else (operator logika), multi-if terpisah, match-case
"""

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. Input Data Pengunjung & String Handling
nama_3025 = input("Masukkan Nama Pengunjung        : ")
umur_3025 = int(input("Input umur anda                 : "))
sim_3025 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]

print()
print("Pilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_3025 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_3025 = int(input("Masukkan jumlah tiket           : "))

# Validasi kelogisan jumlah tiket (if tunggal)
if jumlah_tiket_3025 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

# 2. Pemilihan Wahana menggunakan match-case
nama_wahana_3025 = ""
harga_satuan_3025 = 0
transaksi_valid_3025 = True

match paket_3025:
    case 1:
        nama_wahana_3025 = "Wahana Safari Rimba"
        harga_satuan_3025 = 50000
    case 2:
        nama_wahana_3025 = "Wahana Arung Jeram"
        harga_satuan_3025 = 75000
    case 3:
        nama_wahana_3025 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3025 = 120000
    case 4:
        nama_wahana_3025 = "Wahana Roller Coaster Kilat"
        harga_satuan_3025 = 100000
    case 5:
        nama_wahana_3025 = "Wahana All-Access VIP"
        harga_satuan_3025 = 220000
    case _:
        print("Paket wahana tidak valid!")
        transaksi_valid_3025 = False

if transaksi_valid_3025:
    is_member_3025 = input("Apakah Anda member? (y/t)       : ").strip().lower()
    kode_promo_valid_3025 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

    print()
    print("--- KELAYAKAN PENGENDARA WAHANA ---")

    # 3. Validasi izin kendali wahana (if-elif-else dengan operator logika)
    if paket_3025 == 3:
        if umur_3025 >= 17 and sim_3025 == 'y':
            print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
        elif umur_3025 >= 17 and sim_3025 != 'y':
            print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
        elif umur_3025 < 17 and sim_3025 == 'y':
            print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
    else:
        # Untuk paket selain 3, if-else sederhana cek umur >= 10 tahun
        if umur_3025 >= 10:
            print("Status Akses: Anda memenuhi syarat umur untuk menaiki wahana ini.")
        else:
            print("Status Akses: Anda belum memenuhi syarat umur untuk menaiki wahana ini.")

    # 4. Akumulasi diskon bertingkat (multi-if terpisah)
    subtotal_3025 = harga_satuan_3025 * jumlah_tiket_3025
    total_diskon_persen_3025 = 0

    if subtotal_3025 >= 200000:
        total_diskon_persen_3025 += 10  # Diskon Belanja Besar

    if is_member_3025 in ['y', 'ya']:
        total_diskon_persen_3025 += 5  # Diskon Member

    if kode_promo_valid_3025 in ['y', 'ya']:
        total_diskon_persen_3025 += 15  # Diskon Voucher Promo

    if jumlah_tiket_3025 >= 5:
        total_diskon_persen_3025 += 5  # Diskon Tambahan Rombongan

    # 5. Evaluasi kelulusan audit (if-else)
    nominal_diskon_3025 = subtotal_3025 * (total_diskon_persen_3025 / 100)
    total_bayar_3025 = subtotal_3025 - nominal_diskon_3025

    if total_bayar_3025 > 300000:
        catatan_layanan_3025 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    else:
        catatan_layanan_3025 = "Terima kasih telah berkunjung."

    print()
    print("--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {subtotal_3025:,.0f}")
    print(f"Total Diskon     : {total_diskon_persen_3025}% (Rp {nominal_diskon_3025:,.0f})")
    print(f"Total Bayar      : Rp {total_bayar_3025:,.0f}")
    print(f"Catatan Layanan  : {catatan_layanan_3025}")
    print("Program Selesai")
````

- `.strip()` membuang spasi di awal/akhir input.
- `.lower()` mengubah ke huruf kecil supaya "Y" dan "y" dianggap sama.
- `[0]` mengambil karakter pertama saja, jadi kalau user mengetik "ya" tetap terbaca `'y'`.

```python
    paket_3025 = int(input("Masukkan nomor paket (1-5)      : "))

```
Nomor paket dan jumlah tiket dikonversi ke `int` karena akan dipakai dalam `match-case` dan perhitungan.

```python
    if jumlah_tiket_3025 <= 0:
       print("Peringatan: Kuota tiket tidak valid!")
```
Ini **if tunggal**: hanya mengecek satu kondisi, tanpa `else`. Kalau tiket 0 atau negatif, tampilkan peringatan (program tetap lanjut, sesuai instruksi yang hanya minta "mencetak peringatan").

**2. Pemilihan Wahana (match-case)**
```python
    nama_wahana_3025 = ""
    harga_satuan_3025 = 0
    transaksi_valid_3025 = True
```
Variabel disiapkan lebih dulu (inisialisasi) supaya tidak error jika nanti dipakai di luar blok `match`.

```python
    match paket_3025:
        case 1:
            nama_wahana_3025 = "Wahana Safari Rimba"
            harga_satuan_3025 = 50000
        ...
        case _:
            print("Paket wahana tidak valid!")
            transaksi_valid_3025 = False
```
`match-case` adalah versi Python dari switch-case: mencocokkan nilai `paket_3025` dengan `case 1` s.d. `case 5`. `case _` adalah default — kalau input di luar 1-5, program mencetak pesan error dan `transaksi_valid_3025` diset `False` supaya proses selanjutnya (transaksi) dihentikan.

```python
    if transaksi_valid_3025:
        ...
```
Seluruh proses input tambahan dan perhitungan dibungkus di sini supaya **tidak dijalankan** jika paket tidak valid — ini caranya "menghentikan eksekusi transaksi" tanpa memakai nested-if di dalam match.

**3. Input member & promo**
```python
    is_member_3025 = input("Apakah Anda member? (y/t)       : ").strip().lower()
    kode_promo_valid_3025 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()
```
Disimpan lengkap (tidak dipotong `[0]`) karena nanti dicek dengan `in ['y', 'ya']`, jadi harus menerima "y" maupun "ya".

**4. Validasi izin kendali wahana (if-elif-else + operator logika)**
```python
    if paket_3025 == 3:
        if umur_3025 >= 17 and sim_3025 == 'y':
            print("... boleh mengendarai ATV sendiri.")
        elif umur_3025 >= 17 and sim_3025 != 'y':
            print("... wajib didampingi instruktur.")
        elif umur_3025 < 17 and sim_3025 == 'y':
            print("... Identitas tidak valid...")
        else:
            print("... belum cukup umur...")
```
- Baris terluar (`if paket_3025 == 3`) bukan nested-if terhadap logika bisnis, melainkan sekadar memilih *aturan mana* yang dipakai (aturan ATV vs aturan umum) — sesuai instruksi soal.
- Di dalamnya: `if-elif-else` empat cabang menggunakan operator `and`/`!=` untuk mengecek kombinasi umur dan status SIM sesuai tabel di soal.

```python
        else:
            if umur_3025 >= 10:
                print("... memenuhi syarat umur...")
            else:
                print("... belum memenuhi syarat umur...")
```
Untuk paket selain 3, dipakai `if-else` sederhana: cukup cek `umur >= 10`.

**5. Diskon akumulatif (multi-if terpisah)**
```python
    subtotal_3025 = harga_satuan_3025 * jumlah_tiket_3025
    total_diskon_persen_3025 = 0

    if subtotal_3025 >= 200000:
        total_diskon_persen_3025 += 10
    if is_member_3025 in ['y', 'ya']:
        total_diskon_persen_3025 += 5
    if kode_promo_valid_3025 in ['y', 'ya']:
        total_diskon_persen_3025 += 15
    if jumlah_tiket_3025 >= 5:
        total_diskon_persen_3025 += 5
```
Empat `if` **terpisah** (bukan `elif`) supaya semua kondisi yang terpenuhi ikut menambah persen diskon — inilah maksud "akumulasi/ditumpuk". Kalau dipakai `elif`, hanya satu diskon yang berlaku, padahal soal minta bisa gabungan.

**6. Evaluasi akhir & cetak hasil (if-else)**
```python
    nominal_diskon_3025 = subtotal_3025 * (total_diskon_persen_3025 / 100)
    total_bayar_3025 = subtotal_3025 - nominal_diskon_3025

    if total_bayar_3025 > 300000:
        catatan_layanan_3025 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    else:
        catatan_layanan_3025 = "Terima kasih telah berkunjung."
```
Menghitung nominal diskon (persen dikonversi ke desimal), lalu total bayar. `if-else` menentukan pesan penutup berdasarkan besar total bayar.

```python
    print(f"Subtotal Belanja : Rp {subtotal_3025:,.0f}")
```

f-string dengan format `:,.0f` menampilkan angka dengan pemisah ribuan (koma) dan tanpa desimal, sesuai contoh output di soal (mis. `Rp 240,000`).
